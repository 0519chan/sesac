상품 세부사항 크롤링

- name (상품명)
- price (가격)
- grade (평점)
- review_count (리뷰 개수, 숫자만)
- size_possible (구매가능한 사이즈)
- size_impossible (구매 불가능한 사이즈)
- foot_width (발 볼 넓이)

```python
from selenium import webdriver as wb
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
import pandas as pd
import time
```

```python
driver = wb.Chrome()
driver.implicitly_wait(2)

url = 'https://www.nbkorea.com/product/featuredList.action?cateGrpCode=250100&cIdx=1385'

driver.get(url)

result = {
    'name':[],
    'price':[],
    'grade':[],
    'review_count':[],
    'size_possible':[],
    'size_impossible':[],
    'foot_width':[]
}

img_tags = driver.find_elements(By.CSS_SELECTOR, '#selGoods')

try:

    for i in range(len(img_tags)):
        img_tags = driver.find_elements(By.CSS_SELECTOR, '#selGoods')

        img_tags[i].click()
        
        name = driver.find_element(By.ID, 'displayName').text.strip()
        price = driver.find_element(By.ID, 'dispPrice').text.strip()
        grade = driver.find_element(By.CSS_SELECTOR, 'div.crema_product_reviews_score__container').text.strip()
        review_count = driver.find_element(By.CSS_SELECTOR, 'a.crema-product-reviews-count.crema-applied').text.strip()[1:-1]
        
        size_li_tags = driver.find_elements(By.CSS_SELECTOR, '#optSizeSection ul.items li')
        
        size_p = []
        size_im = []
        for li in size_li_tags:
            if li.find_element(By.CSS_SELECTOR, 'input').get_attribute('disabled'):
                size_impossible = li.find_element(By.CSS_SELECTOR, 'input').get_attribute('data-info')
                size_im.append(size_impossible)
            else:
                size_possible = li.find_element(By.CSS_SELECTOR, 'input').get_attribute('data-info')
                size_p.append(size_possible)
                
        try:
            foot_width = driver.find_element(By.CSS_SELECTOR, '#optWidthSection ul.items li label').text.strip()
            
        except:
            foot_width = '발볼너비 정보 없음'
            
            
        result['name'].append(name)
        result['price'].append(price)
        result['grade'].append(grade)
        result['review_count'].append(review_count)
        result['size_possible'].append(size_p)
        result['size_impossible'].append(size_im)
        result['foot_width'].append(foot_width)
        
        
        driver.back()
        
        
    
    result_df = pd.DataFrame(result)
    print(result_df.shape)
    display(result_df.head())
except Exception as e:
    print(e)
    print(f'{i}번째 제품에서 오류 발생')
    

    
finally:
    driver.quit()
    
    
    
```
