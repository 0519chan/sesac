# 2. BeautifulSoup
- requests를 통해 응답받은 데이터를 BeautifulSoup을 활용하여 html로 형변환을 진행하자 
- html로 형변환된 데이터의 tag를 통해 원하는 데이터에 접근 및 추출해보자 
- 크롤링 순서:
    1. 데이터 요청 및 응답 받기 
    2. 응답 받은 데이터(str)를 html로 형변환 진행 
        - 파이썬의 문자열은 tag를 선택하여 원하는 데이터에 접근할 수 없음 
        - tag에 접근하기위해서는 문자열을 태그에 접근할 수 있는 html로 포맷을 변경해줘야 함 
        - 데이터의 포맷을 변경하는 것을 파싱(parsing)이라고 함 (html로 파싱 가능한 파서의 종류는 html.parser, xlml 등이 있음)
    3. 파싱된 데이터(html)에서 추출하고자 하는 데이터를 담고있는 태그에 접근 
    4. 선택된 태그에서 데이터 추출 `

```python
#!pip install BeautifulSoup4
```

```python
import requests
from bs4 import BeautifulSoup as bs

# 1. 데이터 요청 및 응답받기
response = requests.get('https://crawlingstudy-dd3c9.web.app/01/')

# 2. 응답 받은 데이터(str)를 html로 형변환 진행(parsing, 파싱 과정)
# - bs(파싱할 데이터, 파싱방법)
# - lxml: html로 파싱하는 방법 중 하나

soup = bs(response.text, 'lxml')

print(type(soup))
soup.text
```

```python
h1_tag = soup.select_one('h1')

print(type(h1_tag))
result = h1_tag.text
```

# 3.1 해당되는 태그 모두 갖고오기
- `select(css_selector)`
- `find_all(tag_name)`
- html에서 해당되는 태그는 **모두** 가져와서 list로 return

```python
# 모든 p태그 갖고 오기

import requests
from bs4 import BeautifulSoup as  bs

response = requests.get('https://crawlingstudy-dd3c9.web.app/01/')

soup = bs(response.text, 'lxml')

s_p_tags = soup.select('p')

print(type(s_p_tags))
print(s_p_tags)
```

```python
f_p_tags = soup.find_all(name='p')
print(type(f_p_tags))
print(f_p_tags)
```

# 7. 실습
- 분양 중인 아파트 정보를 크롤링하여 DataFrame을 생성
- 컬럼명은 다음과 같이 정의: 이름, 보증금, 유형, 분양 유형, 세대수, 평형

```python
import pandas as pd
temp = {
    'col1':[],
    'col2':[]
}

pd.DataFrame(temp)
```

```python
url = 'https://crawlingstudy-dd3c9.web.app/03/'

response = requests.get(url)

soup = bs(response.text, 'lxml')

name_list = []
price_list = []
type_list = []
detail_type_list = []
family_count = []
room_size_list = []

name = soup.select('div.sale_box a')


info = soup.select('dl.detail_info dd')

# 이름
for i in range(len(name)):
    name_list.append(name[i].text)
    
# 보증금
for i in range(0, len(info), 4):
    price_list.append((info[i].select_one('strong').text))
    
# 유형
for i in range(1, len(info), 4):
    type_list.append((info[i].text.split('|')[0]))
    
# 분양 유형
for i in range(1, len(info), 4):
    detail_type_list.append((info[i].text.split('|')[1]))

# 세대수    
for i in range(2, len(info), 4):
    family_count.append((info[i].text.split('|')[0]))

# 평형    
for i in range(2, len(info), 4):
    room_size_list.append((info[i].text.split('|')[1]))    
    
    
apart = {
    '이름': name_list,
    '보증금': price_list,
    '유형': type_list,
    '분양 유형': detail_type_list,
    '세대수': family_count,
    '평형': room_size_list
}

pd.DataFrame(apart)
```

```python
url = 'https://n.news.naver.com/mnews/article/016/0002582570'

response = requests.get(url)

soup = bs(response.text, 'lxml')

title = soup.select_one('h2.media_end_head_headline').text

body_list = soup.select_one('#dic_area').find_all(
    string=True, # 텍스트 노드만
    recursive=False # 직접 자식만. 
)

body = "".join(x.strip() for x in body_list)

print(title)
print(body)

```
