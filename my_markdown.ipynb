# 4. 코드 블럭
## 4.1 코드블럭 예시
    코드블럭은 4개의 공벅 또는 한번의 tab으로 들여쓰기를 통해 사용할 수 있음.

## 4.2 줄바꿈 중요
여기는 코드 블럭 밖

    여기는 코드 블럭
        여기는 코드 블럭

여기는 코드 블럭 밖

코드블럭 위아래로 줄바꿈을 하지 않으면
    들여쓰기를 써도
코드 블럭이 생성되지 않음

## 4.3 들여쓰기 없이 코드 블럭 생성
해당 방법을 쓸 땐 코드 블럭 위아래로 줄 바꿈을 쓸 필요가 없음
```
def test_function():
    return "Hello, World!"
```

## 4.4 프로그래밍 언어 강조

\``` 시작점에서 프로그래밍 언어를 명시해주면 해당 언어의 문법 강조가 가능

``` python

    return "Hello, World!"
```
