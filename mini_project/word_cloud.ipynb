```python
import pandas as pd
import glob

# 1. 파일 목록 불러오기 (파일들이 있는 경로 지정)
file_paths = glob.glob('.2025_신용카드이용실적_*.csv')

# 2. 모든 파일을 하나로 합치기
all_data = []
for file in file_paths:
    # 파일 이름에서 '월' 정보 추출 (예: 1월, 2월...)
    month = file.split(' - ')[-1].replace('.csv', '')
    
    df = pd.read_csv(file)
    df['기준월'] = month  # 월 컬럼 추가
    all_data.append(df)

# 통합 데이터프레임 생성
df_total = pd.concat(all_data, ignore_index=True)

# 3. 통합된 파일 저장
df_total.to_csv('2025_신용카드실적_통합.csv', index=False, encoding='utf-8-sig')

print("통합 완료! '2025_신용카드실적_통합.csv' 파일이 생성되었습니다.")
print(df_total.head())
```
