# 1. 조건문 생성
- isinstance() 활용
    - 파이썬에서 어떤 객체가 특정 자료형(클래스)의 객체(인스턴스)인지 확인하는 함수
    - 자료형 검사, 타입 체크에서 많이 쓰이는 파이썬 내장함수
    - isinstance(object, class or tuple) -> 비교할 자료형이 여러개일 경우 튜플로 전달

```python
# 3 > 1 True이면 "정답입니다"를 출력

if(3>1):
    print("정답입니다")
```

```python
x = 3

isinstance(x, int) # 반환값은 항상 bool
```

```python
if isinstance(x, (int, str, float)): # 비교해야할 자료형이 여러 개일 경우 두 번쨰 인자에 tuple로 전달 가능
    print('정수입니다')
else:
    print('No')
print('조건문 실행 완료')
```

# 5. if, elif, else

```python
age = int(input('나이를 입력하세요:'))

if age >= 20 and age <= 29:
    print('20대입니다.')
elif age >= 30 and age <= 39:
    print('30대입니다.')
else:
    print('해당 나이대가 아닙니다')
print('조건문 실행완료')
```

- 특정 값을 입력받은 num변수가 \
2의 배수이면 "2의 배수입니다."를, \
3의 배수면 "3의 배수입니다"를, \
5의 배수이면 "5의 배수입니다"를, \
그 외는 "해당되는 값의 배수가 아닙니다"를 출력하는 조건문을 작성해보자 
- 단, "조건문 실행 완료."는 항상 출력돼야 함

```python
num = int(input())

if num % 2 == 0:
    print('2의 배수입니다')
elif num % 3 == 0:
    print('3의 배수입니다')
elif num % 5 == 0:
    print('5의 배수입니다')
else:
    print('해당되는 값의 배수가 아닙니다')
    
print('조건문 실행 완료')
```

# 8. 중첩   
- if문 안에 또 다른 if문이 들어 있는 구조
- 조건을 여러 단계로 나누어 판단해야 할 때 사용

```python
age = int(input())
is_student = bool(int(input("학생이면 1, 아닐 경우 0을 입력하세요")))

if age >= 20:
    print('성인입니다.')
    if is_student:
        print('학생입니다.')
    else:
        print('성인이지만 학생은 아닙니다.')
else:
    print('미성년자입니다.')
print('조건문 실행 완료')
        



```

- 변수 id와 password에 각각 문자열을 입력받으세요
- id가 "sesac"이고 password가 "1234"이면 "로그인 성공!"을 출력하고, 그렇지 않으면 "비밀번호가 틀렸습니다"를 출력하세요

- 단, "조건문 실행 완료."는 항상 출력돼야 함


```python
id = input('id를 입력하세요')
password = input('password를 입력하세요')

if id == 'sesac':
    if password == '1234':
        print('로그인 성공!')
    else:
        print('비밀번호가 틀렸습니다')
else:
    print('아이디가 존재하지 않습니다.')
print('조건문 실행 완료.')
```

# 실습3 
- A사는 B사에서 C상품 ($10/개) 을 납품합니다. 발주 수량에 따라 수량할인을 고려하여 납품가가 달라집니다.
- 1000개 미만은 수량할인이 적용되지 않습니다.
- 1000개부터 1500개 까지 5% 할인, 1500개 초과 2000개까지 10% 할인, 2000개 초과부터는 20% 할인됩니다.
- 발주 수량을 사용자로부터 입력 받고, 총 발주금액을 출력하는 프로그램을 작성해보세요.
- 소수점은 할인의 일환으로 버립니다.

```python
order = int(input())

if 1000 <= order <= 1500:
    rate = 5
elif 1500 < order <= 2000:
    rate = 10
elif order > 2000:
    rate = 20
else:
    print('수량 할인이 적용되지 않습니다.')
    rate = 0

    
order_amount = 10 * order * ((100-rate)/100)

print(f'적용된 할인율은 {rate}% 입니다.')
print(f'총 발주 금액은 {order_amount}입니다.')
```
