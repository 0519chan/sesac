# 1.3 range() 함수를 활용한 for문 생성
- 반복된 정수를 생성하는 함수 
    - 보통 for문과 가장 많이 함께 사용 됨 
    - 기본 형태: `range(start, stop, step)`
        - start: 시작값 (기본값: 0)
        - stop: 끝값 (끝값-1 까지만 적용되며 실제 끝값은 포함되지 않으며, 필수요소)
        - step: 증가/감소 폭 (기본값: 1)

```python
for i in range(5):
    print(i)
```

```python
for i in range(1,11,2):
    print(i)
```

```python
for i in range(10,-1,-1):
    print(i, end=' ')
```

```python
range(6) # 요소를 확인할 수 없는 객체 형태로 반환이 됨
list(range(6)) # 요소를 확인하고 싶을 때는 iterable 객체로 변환 필요
```

```python
num = int(input())

for i in range(1, num+1):
    print(f'hello, world {i}번')
```

## 1.4 enumerate 함수를 활용한 for문 작성
- enumerate()함수
    - 리스트나 튜플 같은 반복 가능한 객체(iterable)를 순회할 때, 각 요소의 **인덱스**와 **값(value)**을 동시에 반환해주는 파이썬 내장함수
    - 기본 형태: `enumerate(iterable, start=0)`
        - iterable: 리스트, 튜플, 문자열 등 반복가능한 객체
        - start: 인덱스 시작 값(기본은 0)

```python
fruits = ["apple", "banana", "cherry"]

for i in fruits:
    print(i)
    
for i in range(3):
    temp = fruits[i]
    if temp == 'apple':
        print('사과입니다')
    else:
        print('사과가 아닙니다')
```

```python
for idx, value in enumerate(fruits, start=2):
    print('idx: ', idx)
    print('value: ', value)
```

```python
fruits_dic = {}

for idx, value in enumerate(fruits, start=1):
    fruits_dic[idx] = value
    
    
fruits_dic
```

[문제] 1부터 사용자로부터 입렵 받은 자연수까지의 홀수의 합을 구합니다.
- 1부터 입력받은 수까지 홀수 전체를 각각 더하는 산식을 표현하고 
- 그 계산식의 결과를 아래 실행 결과와 같이 출력되도록 구해보세요

[실행 결과] \
입력값: 100 \
1 + 3 + 5 + 7 + 9 + … + 99 = 2500

```python
num = int(input())
total = 1
print(f'입력값은: {num}입니다.')
for i in range(num+1):
    if i == 1:
        print(i, end=' ')
    elif i % 2 != 0:
        total = total + i
        print(f'+ {i}', end=' ')

print('=',total)
        
        
    
```

# 실습

[문제] 학생의 수와 학생들의 성적을 입력 받아 성적 평균을 구하는 프로그램을 구현해보세요. 단, 평균은 정수로 출력하세요.

[실행 결과] 아래와 동일하게 출력하세요.\
학생 수: 3 \
1번 학생의 점수: 77 \
2번 학생의 점수: 88 \
3번 학생의 점수: 99 \
평균: 88점

```python
total_score = 0

std_cnt = int(input('학생의 수를 입력하세요'))

i = 1
while i <= std_cnt:
    score = int(input('점수를 입력하세요'))
    print(f'{i}번 학생의 점수: {score}')
    total_score += score
    i += 1
    
avg_score = total_score/std_cnt
print(f'평균: {int(avg_score)}점')
```

# 4. Continue

```python
for i in range(101):
    if i % 2 == 0:
        continue
    print(i, end=" ")
```

```python
i = 0

while i < 100:
    i += 1
    
    if i % 2 == 0: continue
    
    print(i, end=" ")
```
