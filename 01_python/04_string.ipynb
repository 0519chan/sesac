# 6.문자열과 인덱싱
## 6.1 인덱싱

```python
word = 'python'

print(word[0])
print(word[1])
print(word[2])
print(word[3])
print(word[4])
print(word[5])
```

```python
print(word[-1])
print(word[-2])
print(word[-3])
print(word[-4])
print(word[-5])
print(word[-6])
```

```python
sentence = 'Hello, Word!'

for x in range(len(sentence)):
    print(sentence[x])
```

# 6.2 슬라이싱
## 6.1 증감

```python
word = 'python'

# 0 이상 2 미만
print(word[:2])

# 처음부터 끝
print(word[:])

# 3~ 끝까지
print(word[3:])
```

## 6.2 증감 2 이상

```python
# 0 이상 2 미만
print(word[:2:2])

# 처음부터 끝까지
print(word[::2])

print(word[3::2])
```

## 6.2 역행 (증감이 -1인 경우)

```python
print(word[::-1]) 

print(word[3::-1])
print(word[4:2:-1])
```

# 9. 문자열 내장 함수
## 9.1 len()

- sequence 자료형에서 요소의 개수를 세아려주는 함수
- 문자열에서는 문자열의 길이를 계산 해주는 함수(**공백 포함**)

```python
word = "apple banana"

len(word)
```

## 9.2 replace()
- 특정 문자를 다른 문자로 대체
- 문자열에서 해당되는 문자가 여러개일 경우 모두 대체 문자로 대체됨

```python
word.replace('a','2')

word = word.replace('a','2')

word
```

```python

new_word = word.replace(' ', '')
print(new_word)
print(word)
```

## 9.3 upper(), lower()

- upper(): 소문자를 대문자로 변환
- lower(): 대문자를 소문자로 변환


```python
print(word.upper())
word = word.upper()
print(word.lower())
```

## 9.3 split()
- 특정 패턴을 인자로 전달 시, 해당 패턴을 기준으로 문자열을 분리함
- 분리된 문자들을 list에 담아서 반환

```python
word = 'hello word'


print(word.split())
print(word.split(' '))
print(type(word.split(' ')))
```

[문제] 이름과 점수 3개를 입력받아 아래와 같이 출력하세요

[조건]\
input()함수로 값을 입력받아 f-string을 활용하여 최종 결과를 출력해주세요

[입력 형태] \
이름을 입력해주세요: 홍길동 \
첫 번째 점수를 입력해주세요: 80 \
두 번째 점수를 입력해주세요: 70 \
세 번째 점수를 입력해주세요: 60

[출력 결과]\
저의 이름은 홍길동이고 총점은 210 입니다.

```python
name = input('이름을 입력해주세요: ')
score1 = int(input('첫 번째 점수를 입력해주세요: '))
score2 = int(input('두 번째 점수를 입력해주세요: '))
score3 = int(input('세 번째 점수를 입력해주세요: '))

print(f"저의 이름은 {name}이고 총점은 {score1 + score2 + score3} 입니다")
```
