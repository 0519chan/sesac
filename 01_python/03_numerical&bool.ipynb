# 2. Boolean

## 2.1 비교 연산자

- 아래의 결과를 출력해주세요
    1. 3 > 1
    2. 3 >= 1
    3. 4 < 1
    4. 4 <= 1
    5. 10 == 10
    6. 10 != 10

```python
print(3 > 1)
print(3 >= 1)
print(4 < 1)
print(4 <= 1)
print(10 == 10)
print(10 != 10)
```

```python
# 자료형 확인
print(type(3<1))
```

```python
# 변수를 활용한 비교연산자 결과 출력

x = 10 > 3

print(x)

print(type(x))

x = 1
y = 2

print(x < y)
print(x == y)
print(x != y)

z = x != y
print(z)
```

# 2.2 논리 연산자
## 2.2.1 and
- 둘 다 True여야 True 반환
- 둘 중 하나라도 False이면 False 반환

```python
print(True and True)
print(False and True)
print(False and False)
```

```python
print(type(True and False))
```

```python
# 변수 활용
a = True and False
print(a)
print(type(a))
```

```python
a = True
b = False

print(a and b)
print(type(a and b))
```

### 2.2.2 OR
- 둘 중 하나만 True여도 True 반환

```python
print(True or False)
print(False or False)
```

### 2.2.3 NOT



```python
print(not False)
print(not True)
```

## 2.3 비교 + 논리 연산자

```python
(5 > 3) and (2 < 4)
```

```python
(5 == 5) or (10 != 10)
```
