# 3. 인덱싱과 슬라이싱

## 3.1 1차원

### 3.1.1 인덱싱

```python
fruit = ['apple', 'banana', 'cherry']

for i in range(len(fruit)):
    print(fruit[i])

#역순

for i in range(len(fruit), -1):
    print(fruit[i])
```

### 3.1.2 슬라이싱

```python
# 모든 요소 출력
print(fruit[:])

# banana, cherry 출력
print(fruit[1:])

# cherry 출력
print(fruit[2:])

# step 활용
a = ['this', 'is', 'a', 'simple', 'list']

# is, simple 출력
a[1::2]
```

# 4. 할당과 복사

list가 다차원일 경우 할당과 얕은 복사, 깊은 복사를 통한 차이점을 파악

### 4.1.1 할당

```python
a = [1, 2, 3, 4, 5]
b = a # 할당

print('a:', a)
print('id a :', id(a))
print('b:', b)
print('id b :', id(b))

b[0] = 0

print('a:', a)
print('b:', b)

# 동일한 메모리 주소를 참조하고 있기 때문에 둘 중 하나의 값이 변경되면 모두 반영이 됨
```

### 4.1.2 복사

```python
c = a.copy() # 얕은 복사

print('a : ', a)
print('c : ', c)

print('\nid a:', id(a))
print('id c:', id(c))

c[0] = 1

print('\na : ', a)
print('c : ', c)
```

## 4.2 2차원의 할당과 복사
### 4.2.1 할당

```python
a = [[1,2], [3,4], [5,6]]

b = a

print('a: ', a)
print('b: ', b)

print('\nid a', id(a))
print('id b', id(b))

b[0][0] = 100

print('a: ', a)
print('b: ', b)
```

```python
c = a.copy()

print('a: ', a)
print('c: ', c)

print('\nid a', id(a))
print('id c', id(c))

print('\nid list a', id(a[0]))
print('id list b', id(b[0]))

c[0][0] = 5 # 2차원 데이터 변경
c[1] = [100, 200] # 1차원 데이터 변경

print('\na: ', a)
print('c: ', c)
```

### 4.2.3 깊은 복사

```python
import copy

a = [[1,2], [3,4], [5,6]]
b = copy.deepcopy(a) # 깊은 복사

print(a)
print(b)

print('\nid a', id(a))
print('id b ', id(b))

print('\nlist a id', id(a[0]))
print('list b id', id(b[0]))

b[0] = [100, 200]
b[1][0] = 300

print('\n',a)
print(b)
```

# 5. 리스트의 다양한 기능
## 5.1 슬라이스 요소 할당 및 삭제

```python
a = [1,2,3,4,5,6,7]

# 인덱스 1~3 요소를 'a','b','c'로 변경
a[1:4] = ['a', 'b', 'c']

a

# 방법1) 인덱스 1~3 요소들을 삭제 

a[1:4] = []

a

# 방법2) 인덱스 1~3 요소들을 삭제
del a[1:]

a
```

## 5.4 요소 추가
### 5.4.1 append()

- 리스트의 마지막에 새로운 요소(객체)를 추가

```python
a = [1, 2, 3]
b = [4, 5, 6]
c = (100, 200, 300)

a.append(4)
print(a)

a.append(b)
print(a)

a.append(c)
print(a)

print(len(a))

# append는 결과가 원본에 바로 반영되기 때문에 실행할 때 주의
```

### 5.4.2 extend()

리스트의 마지막에 추가하는 iterable 객체의 요소들을 unpacking해서 추가(확장의 개념)

```python
a = [1, 2, 3]
b = [4, 5, 6]
c = (100, 200, 300)
d = {'key': 'value'}

a.extend(b)
a

a.extend(c)
a

a.extend(d)
a
```

### 5.4.3 insert()

원하는 인덱스 위치에 원하는 데이터를 추가 (원래 위치에 있던 요소는 뒷 순서로 밀려남)

```python
a = [1, 2, 3]

a.insert(0, 4) # insert(인덱스, 추가할 요소)
a

a.insert(2,100)
a
```
