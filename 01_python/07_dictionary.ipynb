# 2. 딕셔너리의 특징
## 2.1 key는 고유해야함

```python
people = {'name':'Tom', "name":"Amy"}

print(people) # 중복된 key 값이 있을 경우 가장 마지막에 추가된 key-value 값이 반환됨
```

## 2.2 key가 될 수 있는 자료형
- 사용 가능한 자료형(가변성 X)
    str, int, float, bool, tuple
- 사용 불가능한 자료형(가변성 O)
    list, dict, set

```python
my_dict = {
    "key":"vlaue",
    1:"value2",
    1.1:"value3",
    True:"value4",
    (1,2,3) : "value5"
}

my_dict
```

```python
my_dict = {
    "key":"vlaue",
    1:"value2",
    1.1:"value3",
    True:"value4",
    (1,2,3, [1,2]) : "value5"
   # [{1, 2, 3}] : "value6"
}

my_dict

# hashable : 객체가 변하지 않는 고유한 값(해시값)을 가짐
# list, dict, set는 가변성의 특징을 갖고 있기 때문에 dict의 key로 쓸 수 없음
```

##  2.3 value가 될 수 있는 자료형
- 모든 자료형이 value가 될 수 있음

```python
my_dict = {
   1: "value",
   2: 2,
   3: 3.3,
   4: True,
   5: (1,2,3),
   6: ['a','b','c'],
   7: {1,2,3},
   8:{1:'a'}
}

my_dict
```

```python
type(my_dict)

my_dict.keys()
my_dict.values()

list(zip(list(map(type, my_dict.keys())), my_dict.keys()))
```

[문제] 이름, 나이, 연락처를 받아 딕셔너리를 생성하여 출력해주세요. 

[실행] \
이름을 입력해주세요: 홍길동 \
나이를 입력해주세요: 27 \
연락처 입력해주세요: 010-1234-6789 

[결과] \
{'이름': '홍길동', '나이': '27', '연락처': '010-1234-6789'}


```python
name = input('이름을 입력해주세요')
age = input('나이를 입력해주세요')
phone = input('연락처를 입력해주세요')


info = {'name':name, 'age':age, 'phone':phone}

print(info)
```

```python
name, age, phone = input().split(',')

info = {'name':name.strip(), 'age':age.strip(), 'phone':phone.strip()}


print(info)
```

```python
keys = ['이름', '나이', '연락처']
values = input().split(',')

res = dict(zip(keys, values))

res
```

# 8. 실습2

[문제]
input() 함수를 통해 원하는 과일 이름을 입력하면 가격이 출력되는 프로그램을 구현해봅시다.
1. 과일 이름(str)을 key로 만들고 가격(int)을 value로 해서 2번의 자료로 딕셔너리 자료형을 만듭니다. 
2. 사과는 1000원, 바나나는 700원, 오렌지는 1500원, 파인애플은 2000원
3. 사용자로부터 과일 이름을 입력받습니다. 
4. 해당되는 과일의 가격을 출력해주세요.

[실행 결과] \
과일 이름을 입력하세요: 사과 \
사과의 가격은 1000원입니다.

```python
keys = ['사과', '바나나', '오렌지', '파인애플']
values = [1000, 700, 1500, 2000]

info = dict(zip(keys, values))

names = input('과일 이름을 입력하세요:')

print(names + '의 가격은', str(info[names]) + '원입니다.')
print(f"{names}의 가격은 {info[names]}원입니다.")
```
