[문제]
- 정수를 입력했을 때 짝수면 "짝수", 홀수면 "홀수" 문자열을 반환하는 함수를 생성해보자
    - Docstring 작성
    - 함수명은 odd_even으로함
    - 출력 항목은 같은 셀에 작성해주세요. 출력 항목에 대한 내용이 모두 출력돼야 합니다.

[출력 항목]
1. odd_even.\_\_doc\__
2. odd_even(3)
3. odd_even(112)

[출력 결과]
1. 사용자가 정의한 docstring 원문 
2. 홀수
3. 짝수 

```python
def odd_even(a):
    """
    int a를 입력받아 2로 나누어 떨어지면 짝수 아니면 홀수인지를 판별

    Args:
        a(int) : 사용자가 전달하는 임의의 정수
        
        
    return:
        str : '짝수' or '홀수'
    """
    
    result = ''
    if a % 2 == 0:
        result = '짝수'
    else:
        result ='홀수' 
        
    return result


```

```python
print(odd_even.__doc__)
print(odd_even(3))
print(odd_even(112))
```

## 위치 전용 인자 

```python
# 위치 전용으로 명시하지 않은 경우
def not_position(temp):
    print(temp)
    
# 위치 전용으로 명시
def position_only(temp, /):
    print(temp)
```

```python
not_position(42) # 위치
not_position(temp=42) # 키워드
 
position_only(42) # 위치
position_only(temp=42) # 키워드
```

## 2.5 키워드 전용 인자


```python
def not_keyword(temp):
    print(temp)
    
def keyword(*, temp):
    print(temp)
    
    
print(not_keyword(42))
print(not_keyword(temp=42))

print(keyword(temp=42))
print(keyword(42))
```

# 3. 데코레이터(Decorator)

```python
def decorator(func): # 수정할 함수를 인자로 받음
    def wrapper(): # 수정할 함수를 감싸는 함수 -> 최종적으로 반활될 함수
        print('함수 실행 전 작업 구간')
        func() # 수정할 함수 호출
        print('함수 실행 후 작업 구간')
    return wrapper
```

```python
@decorator
def test_func():
    print('수정할 함수(기존의 함수)')
```

```python
def logger(func):
    def wrapper(*args, **kargs):
        print(f'실행중인 함수: {func.__name__}')
        return func(*args, **kargs)
    return wrapper


@logger
def say_hello(name):
    print(f'안녕하세요 {name}님!')
    
say_hello('chan')
```
