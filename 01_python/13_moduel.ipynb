# 1. 모듈

## 1.1 모듈만 import 및 실행

```python
import calc

# 처음 import만 모듈 스크립트 파일이 실행되고, 그 이후에는 모듈이 캐싱되어(저장) 캐싱된 모듈을 들고 오기 때문에
# 두번째 실행부터는 모듈 스크립트 파일이 실행되지 않음
# 변경사항을 반영하기 위해서는 커널을 다시 시작하여 변경된 모듈을 캐싱시켜줘야함.
# 캐싱: 자주 사용하는 데이터나 결과를 빠른 저장소에 미리 저장해서 다음 요청을 더 빠르게 처리하기 위한 기술
```

```python
calc.name
```

```python
import calc

print(calc.add(3,4))
print(calc.sub(3,4))
print(calc.dup(3,4))
print(calc.div(3,4))
```

```python
# 함수까지 import한 경우 -> 에러 발생
# import에서는 모듈에서 클래스나 함수까지 접근할 수 없음
import calc.add
```

```python
from calc import add, sub, name

print(add(1,2))
print(sub(5,1))
name
```

# 2. 패키지

## 2.1 패키지 import

- 패키지만 import할 경우 패키지 내부 모듈에 접근 할 수 없음
    - 이럴 경우엔 패키지 안의 모듈도 같이 import 해줘야함
- 모듈을 같이 import 하지 않고 패키지 import만으로 모듈에 접근하고 싶을 경우 패키지 내에 \_\_init\_\_ 파일을 생성해서 활용해야함

```python
import my_package

my_package.area # 패키지만 import하면 모듈로 접근 할 수 없음 -> init 파일 생성 후 다시 실행
```

# 2.2 패키지에서 함수로 접근

```python
# import만으로는 패키지에서 클래스나 함수까지는 접근할 수 없음

import my_package.area

my_package.area.circle(3)
```

```python
# from을 통해 불러올 함수의 출처를 명시해준 후 import로 함수를 불러와야함
from my_package.area import circle

circle(3)
```

# 3. as 예약어
- as 예약어를 통해 import할 기능들의 이름의 별칭을 생성 및 활용 할 수 있음

```python
from my_package.area import circle as cir

cir(3)
```

# 4. 패키지와 모듈 위치 찾기

- sys: 파이썬 표준 라이브러리에 포함된 시스템 관련 기능을 제공하는 모듈로, 파이썬 인터프리터 자체와 운영체제 환경과 상호작용할 수 있도록 도와주는 모듈

    3. site-packages (설치된 패키지 디렉터리)
    4. 표준 라이브러리 설치 경로
