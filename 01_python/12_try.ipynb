# 예외처리

# 1. try & except

```python
print(10/0)
print('에러 발생')
```

```python
try:
    print(10/0)
except:
    print('에러발생')
```

```python
try:
    x = int(input('숫자를 입력하세요: '))
    print(10/x)
except ZeroDivisionError:
    print('0으로 나눌 수 없습니다.')
except ValueError:
    print('유효한 숫자를 입력하세요.')
```

```python
x = range(4)

try:
    for i in x:
        y = 10/x[i]
except ZeroDivisionError:
    print('0으로 나눌 수 없습니다.')
except IndexError:
    print('인덱스 에러 발생')
```

```python
x = range(1, 4) # 1, 2, 3

try:
    for i in x: 
        y = 10/x[i]
        print(y)
except ZeroDivisionError:
    print('0으로 나눌 수 없습니다.')
except IndexError:
    print('인덱스 에러 발생')
```

```python
try:
    x = int(input('숫자를 입력하세요'))
except Exception as e:
    print(f'오류가 발생: {e}')
```

```python
try:
    x = int(input())
    print(10/x)
except ZeroDivisionError as e:
    print(e)
else:
    print('성공적으로 실행')
finally:
    print('프로그램이 종료')
```
