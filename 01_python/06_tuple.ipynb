# 튜플(tuple)

# 1. 튜플 생성

```python
a = 1, 2, 3
b = (4,5,6)

print(a)
print(b)
print(type(a))
print(type(b))
```

```python
a = ("하나", 123, True, 45.7, [1,2], (4, 5), {'key':'value'})

print(a)
print(type(a))

# 튜플이 아닌 것

a = 1
print(type(a))

b = (1) # 쉼표가 없을 경우 단순히 정수를 괄호에 감싼 형태가 됨(우선순위를 나타내는 괄호로 인식)
print(type(b)) 

c = (1,)
print(type(c))

d = 1,
print(type(d))
```

# 2. 불변의 객체 튜플

```python
a = (1, 2, 3 , 4)
b = [1, 2, 3, 4]

# 요소 변경
b[0] = 100
b

a[0] = 100 # 튜플은 값을 변경이 불가능
a

```

```python

# 요소 삭제

del b[0]

b

del a[0]
a
```

# 3. 유용한 파이썬 함수들

## 3.1 zip()


```python
keys = ["name", "age", "city"]
value = ["chan", "27", "seoul"]

mapped = zip(keys, value) # generator 객체가 반환(메모리에서 출력될 준비만 된 상태)

#list(mapped)
tuple(mapped) # zip 객체는 한 번 소모되면 재사용 불가

    
    
```

```python
mapped = zip(keys, value)

dict(mapped)
```

## 3.2  map()


```python
a = [1.2, 2.5, 3.7 , 4.1]

mapped = map(int, a)

print(mapped)
print(list(mapped))
print(type(mapped))
```

## 3.3 input, split, map을 같이 활용해보기

```python
num = input("정수 4개를 입력하세요")
```

```python
num_split = num.split(' ')
num_mapped = list(map(int, num_split))
```

```python
print(type(num_split[0]))
type(num_mapped[0])
```

국어, 영어, 수학, 과학 점수를 한 번에 입력 받습니다. \
이를 kor, eng, mat, sci라는 변수에 각각 저장하고, 사칙 연산을 활용하여 평균 점수를 산출해보세요 \

[실행 결과]
- 국어, 영어, 수학, 과학 점수를 입력해주세요: 98,100,96,100
- 평균 점수는 98.5입니다.

```python
score = input('점수를 입력해주세요')
```

```python
score_mapped = list(map(int, score.split(',')))

kor, eng, mat, sci = score_mapped[:]

(kor + eng + mat + sci)/4
```
