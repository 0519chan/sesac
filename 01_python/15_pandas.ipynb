# 1. Series
- 각 데이터에 라벨(label)이 붙은 1차원 데이터

## 1.1 Series 생성

```python
import pandas as pd

s = pd.Series([9_904_312, 3_448_737, 2_890_451, 2_466_052],
          index=['서울', '부산', '인천', '대구'])
```

```python
print(s.index)
print(s.values)
print(type(s.index))
print(type(s.values))
```

```python
# index.name으로 인덱스에 이름 붙이기

s.index.name = 'city'
s
```

```python
sereis = pd.Series(['하나', '둘', '셋' ,'넷' ,'다섯'], index = [_ for _ in range(1,6)])

sereis
```

# dict를 활용하여 series 생성


```python
d = {'a':1, 'b':2, 'c':3}

s2 = pd.Series(d)

s2
```

```python
# index 변경

s2.index = ['A', 'B', 'C']
s2
```

```python
# 인덱스 없이 시리즈 생성

s3 = pd.Series(range(5))

s3
```

## isNull

```python
s1 = pd.Series([9_904_312, 3_448_737, 2_890_451, 2_466_052],
          index=['서울', '부산', '인천', '대구'])

s2 = pd.Series([9631482, 3393191, 2632035, 1490158],
              index=['서울', '부산', '인천', '대전'])


s_new = (s1-s2)/s2

s_new.isnull().sum()
```

```python
s_new.notnull()
```

## 2.1 DataFrame 생성
