#자료형: 다양한 유형의 데이터를 알아보자

- 아래의 클래스들와 함수를 활용하여 데이터의 유형을 확인해보자
    - type(`type()`): 데이터의 유형을 봔환해주는 클래스
    - print(`print()`): 메모리에 저장 돼 있는 값을 불러 콘솔에 출력해주는 함수

# 1. 수치형
## 1.1 int: 정수

```python
type(42)
type(1528)
```

```python
print(type(42))
print(type(12))
type(4)
print(type(2242))
print(type(32))
```

## 1.2 float: 실수

```python
type(1.254)
```

```python
print(type(42.))
```

# 2. Sequence
## 2.1 Str: 문자열

```python
'"Hello World!"'
```

```python
'Hello World!'
```

```python
type('Hellow World!')
type("Hellow World!")
```

```python
print(type('Hellow World!'))
print(type('"Hellow World!"'))
```

## 2.2 List

```python
print(type([3,4,4.12, 'Hellow']))
print(type([1]))

print(type([(10,20), 'a']))
```

## 2.3 Tuple

```python
(1,2,3, 'Coffee',['A', 0.2])
```

```python
type([1,2, ('ABC', 'AAA', 0.3)])
```

```python
type((1,2,3, 'Coffee',['A', 0.2]))
```

```python
type((3,4, (10, 'A')))
```

```python
type(['a', 1, [(3,4), 2.1]])
```

# 3. Mapping : dict(dictionary)

key:value 쌍으로 구성 되어 있는 데이터 타입

```python
{'name': 'yoon', 'age':27}
```

```python
print(type({'name': 'yoon', 'age':27}))
```

# 4. Boolean(Bool)

참과 거짓 두가지의 값만 가질 수 있는 데이터 타입


```python
True, False
```

```python
print(type(True))
print(type(False))
```

# 5. Set

중괄호 {}에 한 개 이상의 데이터가 묶여 있는 형태

```python
type({3,4, 'abc', (1, 2)})

type({(10,20)})
```

```python
print('1. :', type(15827))
print('2. :', type(7.0))
print('3. :', type({1:'a'}))
print('4. :', type("Hi, My Name is 'Dasiy'"))
print('5. :', type((812.324992, 'orange', ('test'))))
print('6. :', type(['apple', 3.513, {'test':1}]))
```
