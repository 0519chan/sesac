# 클래스(class)

# 1. class 정의와 객체 생성

```python
class Dog:
    def __init__(self, name, age): # "클래스"에서 받아들이는 파라미터 정의
        # 클래스의 속성(데이터) 생성
        # 클래스의 속성: 클래스 내부에서 사용되는 변수
        self.name = name
        self.age = age
```

```python
buddy = Dog(name='buddy', age=2)

print(buddy.name)
print(buddy.age)
```

```python
max = Dog('max', 4)

print(max.name)
print(max.age)
```

# 2. 메서드(Method)

- 메서드는 객체의 기능을 의미

```python
class Dog:
    def __init__(self, name, age): # "클래스"에서 받아들이는 파라미터 정의
        # 클래스의 속성(데이터) 생성
        # 클래스의 속성: 클래스 내부에서 사용되는 변수
        self.name = name
        self.age = age
        
    def bark(self):
        return '멍멍'
        
```

```python
buddy = Dog('buddy', 2)
```

# 3. 속성(Attribute)
- 속성은 객체의 데이터를 의미

```python
class Dog:
    def __init__(self, name, age):
        # 인스턴스 속성: 각 객체에 독립적으로 적용됨
        self.name = name
        self.age = age
        self.stress = 0
        
    def bark(self):
        self.stress += 1
        return f'{self.name}가 짖습니다. 멍멍'
```

```python
buddy = Dog('buddy', 2)

print(buddy.name)
print(buddy.age)
print(buddy.stress)
print(buddy.bark())
print(buddy.stress)
```

```python
tom = Dog('tom', 12)

print(tom.name)
print(tom.age)
print(tom.stress)
print(tom.bark())
print(tom.stress)
```

# 5. 상속
## 5.1 상속
- 자식 클래스 정의 시 부모 클래스를 상속 받아 부모 클래스의 기능을 활용해보자

```python
class Animal:
    def __init__(self):
        self.hungry = 0
        
    def eat(self):
        self.eat -= 10
        print('밥 먹음 ->', self.hungry)
        
    def walk(self):
        self.hungry += 10
        print('산책함 ->', self.hungry)
        
        
class Dog(Animal):
    def __init__(self):
        super().__init__()
        
    def sound(self):
        print('멍멍')
        
class Cat(Animal):
    def __init__(self):
        super().__init__()
        
    def sound(self):
        print('야옹')
```

```python
print('개----------')
buddy = Dog()
print(buddy.hungry)
buddy.sound()
buddy.walk()
buddy.walk()

print('최종 배고픔 상태:', buddy.hungry)

print('\n고양이------------')

nabi = Cat()
print(nabi.hungry)
nabi.sound()
nabi.walk()
nabi.walk()
```

```python
class Character:
    def __init__(self):
       self.health = 200
    
    def move(self):
        self.health -= 10
        print("운동 ->", self.health)
        
    def rest(self):
        self.health += 10
        print('휴식 ->', self.health)
        
    def check_health(self):
        print("현재 상태 -> ", self.health)
        

    
```

```python
chan = Character()

chan.health
chan.move()
chan.rest()
chan.check_health()
```

# 7. 실습2

[문제]
- Knight 와 Healer 클래스는 Charcter 클래스를 상속
- Knight 클래스의 move() 호출 시
    - health -5 
    - "**추가 데미지 -5"와 "현재 상태 -> {health)"를 출력
- Knight 클래스 attack() 호출 시
    - "공격합니다"를 출력

- Healer 클래스 mana속성에 100 할당
- Healer 클래스의 heal() 메서드를 추가하고, character인자를 받습니다 (이 때 character는 Character클래스로 생성된 객체입니다.)
- Healer 클래스의 heal() 메서드 호출 시
    - "치료합니다" 출력
    - mana -10, 
    - 전달받은 인자인 character객체의 rest() 메소드를 호출
- Healer 클래스에 현재 마나속성을 확인할수있는 check_mana() 메서드 추가
- Healer 클래스의 check_mana() 호출 시 "현재 남은 mana -> {mana}"를 출력해주세요

```python
class Knight(Character):
    def __init__(self):
        super().__init__()
        
    def move(self):
        super().move()
        self.health -= 5
        print('추가 데미지 -5, 현재 체력', self.health)
        
    def attack(self):
        print('공격합니다.')
        
        
class Healer(Character):
    def __init__(self):
        super().__init__()
        self.mana = 100
        
    def heal(self, character):
        super().move()
        print('치료합니다.')
        self.mana -= 10
        character.rest()
    
    def check_mana(self):
        print('현재 남은 mana', self.mana)
    
```

```python
knight = Knight()

knight.move()
knight.attack()

healer = Healer()

healer.mana
healer.heal(character=knight)
healer.check_mana()
```
