# 1. 변수 생성

다양한 유형의 데이터를 변수에 할당

```python
# int 할당
x = 10

print(x, type(x))

# float 할당
y = 0.345

print(y, type(y))

# string 할당
c = 'Yoon'

# bool 할당
d = True

# list 할당
fruits = ['apple', 'banana', 'cherry']

# tuple 할당
coordinates = (10.0, 20.0)

# dict 할당
person = {'name':'yoon', 'age':27}

# set 할당
unqiue_number = {2, 3, 5, 7}

```

```python
print(x, type(x))
print(y, type(y))
print(c, type(c))
print(d, type(d))
print(fruits, type(fruits))
print(coordinates, type(coordinates))
print(person, type(person))
print(unqiue_number, type(unqiue_number))
```

# 4. 동시 할당
- 여러 개의 변수를 동시에 선언하는 방식
- 대입 연산자를 기준으로 좌변에는 여러 개의 변수, 우변에는 리터럴 자료형이 위치하며 리터럴의 요소 개수는 좌변의 변수 개수와 일치해야함
- 좌변에 있는 변수의 개수와 우변에 있는 리터럴 요소의 개수가 일치하면, 각각의 변수에 맅터럴 내의 요소들을 순서대로 할당

```python
a, b = 1, 20 # 소괄호가 생략된 튜플의 요소들이 언패킹 되어 각각의 변수에 할당됨
```

```python
print(a, b)
```

```python
x, y, z = [10,20,30]

print(x)
print(y)
print(z)
```

```python
x,y,z ='abc'

print(x)
print(y)
print(z)
```

# 5. 비어 있는 변수

- None : 아무 값도 없는
    - 값이 없음을 나타내는 특별한 상수
    - None Type이라는 고유한 타입을 가짐
    - 논리 연산(Bool)에서는 False로 간주
    - 변수나 함수에서 반환할 값이 없거나 값이 존재하지 않을 때 None을 사용
    - 다른 프로그래밍이나 언어에서는 None 대신 Null이 쓰임

```python
x =None
```

```python
type(x)
```

# 6. 변수 삭제
- del 예약어를 활용하여 x에 할당된 데이터를 삭제해보자
- 메모리에서 변수에 할당된 데이터를 삭제하여 변수가 더 이상 해당 주소를 참조(접근)할 수 없도록 함.

```python
x = 42

print(x)
print(id(x))

del(x)
x
```

```python
x = None

print(id(x))

x = 1

print(id(x))
```
