# 1. sidebar & columns
- with구문
    - with 구문은 특정 블록 안에서만 유효한 설정이나 리소스를 관리하기 위해 사용 됨
    - 즉, 요소를 하나의 "영역"으로 묶을 때 쓰임
- sidebar: 왼쪽 사이드 바 설정
- columns: 컬럼 공간 설정

```python
%%writefile my_web/06_1_layout.py
import streamlit as st
import time

# -------------------------------side bar------------------------------------------#

with st.sidebar:
    st.title('야식 선택')
    multi_select = st.multiselect('오늘 야식은?', options=['치킨','피자','햄버거'])
    multi_select = ", ".join(multi_select)
    st.write(f"결과: {multi_select}")
    
    
"---"

st.title('spinner')
if st.button('진행'):
    with st.spinner('데이터 전달 중 ...'):
        time.sleep(3)
        st.success('전달 완료')
        
        
# -------------------------------columns------------------------------------------#

col1, col2, col3 = st.columns(3)

with col1:
    st.header('칼럼1', divider = 'red')
    st.image('../../data/hamster.jpg', caption='cute')

with col2:
    st.header('칼럼2', divider = 'green')
    st.image('../../data/hamster.jpg', caption='cute')
    
with col3:
    st.header('칼럼3', divider = 'rainbow')
    st.image('../../data/hamster.jpg', caption='cute')
    
    
```

# 2. tab

- 탭 공간 설정 (특정 탭을 선택하면 화면에 해당 탭의 내용만 보임)

```python
%%writefile my_web/06_2_layout.py
import streamlit as st

tab1, tab2 = st.tabs(['좌측 탭', '우측 탭'])


with tab1:
    col1, col2, col3 = st.columns(3)
    
    with col1:
        st.header('칼럼1', divider = 'red')
        st.image('../../data/hamster.jpg', caption='cute')

    with col2:
        st.header('칼럼2', divider = 'green')
        st.image('../../data/hamster.jpg', caption='cute')
        
    with col3:
        st.header('칼럼3', divider = 'rainbow')
        st.image('../../data/hamster.jpg', caption='cute')
    
```

# 3. expander & empty & container 

- expander: 사용자가 클릭하여 확장하거나 축소할 수 있는 섹션 설정
- empty: 비어있는 자리표시자(placeholder)
    - 페이지의 특정 위치에 하나의 요소를 출력할 수 있는 독립적인 공간 설정
    - 새 요소를 추가하면 기존의 요소가 대체되므로 가장 최신의 업데이트를 출력하는 데 포커스 맞춤
    - 단일 콘텐츠를 동적으로 업데이트하고 표시하는 데 유용함
- container: 특정 위치에 여러개의 요소를 출력할 공간 설정
    - 컨테이너는 empty와 달리 기존내용을 자동 초기화 하지 않고 덮어쓰기처럼 작동 
    - 새 요소를 추가하면 기존 요소를 바꾸지 않고 단순히 컨테이너에 추가 -> 누적된 정보 집합을 출력하는 데 포커스 맞춤 
    - 컨테이너는 영구적인 UI 공간
        - empty와 달리 재사용 가능한 공간
        - 그 안에 출력된 요소들은 자동으로 초기화되지 않음 
        - 단순히 cont.container()를 새로 호출해도 기존 요소는 즉시 지워지지 않음 
        - cont.container()는 새로운 하위 컨테이너를 생성하는것이지 기존 컨텐츠를 제거하지 않음 
    - [Streamlit의 렌더링 방식]
        - Streamlit은 화면을 프레임단위로 리렌더링
        - 버튼 클릭 시 전체 스크립트를 다시 실행하지만, 브라우저에는 변화된 부분만 점진적으로 반영(diff patching) 됨
        - 특히 컨테이너 안에 있던 텍스트가 겹치거나 재렌더링될 경우, CSS/JS가 그것을 fade-out 처리하거나 잠깐 유지했다가 제거하게 되는 것처럼 보일 수 있음 
        - 완전히 초기화를 하고싶을 경우 cont.empty()를 활용해야 함 


```python
%%writefile my_web/06_03_layout.py
import streamlit as st
import time

# -------------------------------Expander------------------------------------------#
st.subheader('Expander')

with st.expander('Expander 공간 1'):
    st.write('햄스터1')
    st.image('../../data/hamster.jpg', width = 300)
    
    
with st.expander('Expander 공간 2'):
    st.write('햄스터2')
    st.image('../../data/hamster.jpg', width = 300)
    
    
""
"---"
# -------------------------------empty------------------------------------------#
st.subheader('empty')

empty = st.empty()
empty.write('여기는 empty 공간입니다.')

c1, c2 = st.columns(2)
start_btn = c1.button('시작', key='empty_start')
clear_btn = c2.button('초기화', key='clear_start')

# 시작 버튼 클릭 시, empty 공간에 카운트다운이 1초 간격으로 진행
if start_btn:
    with empty:
        for i in range(6):
            cnt = 5 - i
            st.write(f":red[{cnt}]초 남았습니다.")
            time.sleep(1)
        st.write('empty 내부 카운트다운 실행 완료')
        
        
if clear_btn:
    empty.empty()
    empty.write('초기화 중입니다.')
    time.sleep(2)
    
    empty.write('empty 공간 초기화 완료')
    time.sleep(2)
    
    empty.write('empty.write() 작성 예시입니다.')
    st.write('st.write() 작성 예시입니다.')
    
    
""
"---"
# -------------------------------container------------------------------------------#
st.subheader('container')

cont = st.container()
cont.write('여기는 container 공간입니다.')

# container 외부 공간
c1, c2 = st.columns(2)

start_btn_c = c1.button('시작', key='const_start')
clear_btn_c = c2.button('초기화', key='const_clear')

if start_btn_c:
    with cont:
        for i in range(6):
            cnt = 5 - i
            st.write(f":red[{cnt}]초 남았습니다.")
            time.sleep(1)
        st.write('container 내부 카운트다운 실행 완료')
        
        
if clear_btn_c:
    cont.container()
    cont.write('초기화 중입니다.')
    time.sleep(2)
    
    cont.write('empty 공간 초기화 완료')
    time.sleep(2)
    
    cont.write('empty.write() 작성 예시입니다.')
    st.write('st.write() 작성 예시입니다.')
```
