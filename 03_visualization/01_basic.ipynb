# Streamlit
[다양한 텍스트 출력 함수]
- st.title: 타이틀 - 대
- st.header: 타이틀 - 중
- st.subheader: 타이틀 - 소
- st.text: 일반 텍스트
- st.markdown: 마크다운 출력(HTML/CSS 코드로 일부 커스터마이징 가능)
- st.write: 텍스트나 파이썬 변수/객체 출력
- st.latex: 수식 출력
- st.caption: 짧은 설명문(캡션)출력

[기타]
- %%: 주피터노트북 매직 명령어

    - 단, 매직 명령어 사용 시 해당 코드의 위쪽에 다른 코드나 주석이 있으면 동작하지 않음

```python
%%writefile my_web/01_text.py
import streamlit as st
import pandas as pd

# 타이틀 출력"
st.title('지금 아주 졸려요 - 대')

st.header('조금 깼어요 - 중')

st.subheader('목이 아파요 - 소')
st.subheader('도넛 뭐 먹지 - 소', divider=True)
st.subheader('금액은?', divider='red' )

"" # 공백 추가

# 일반 텍스트 출력 (단순 텍스트 출력용)
st.text("간단한 텍스트 출력")

# 마크다운 출력
st.markdown("*마크다운* 출력") # italic 적용
st.markdown('**마크다운**') # bold 적용
st.markdown('[마크다운](https://blog.naver.com/anthouse28/221210836545)')
st.markdown(':smile: :100: :ocean:') # 이모지 적용

st.markdown('마크다운 :red[**색상**] 적용')
st.markdown('마크다운 :blue[**색상**] 적용')

'마크다운 :green[**색상**] 적용'

""

# 파이썬 객체 출력
x, y = 1, 2

st.write(1,2)
st.write('x= ',x, "그리고", 'y= ',y)
st.write(f'x={x}, y={y}') # 문자로 인식
# st.text('x=',x) -> text 메소드는 str값만 전달 가능하며 추가적인 매개변수 또는 파이썬 객체는 전달 불가능

st.latex('E = mc^2')
st.latex(r'Area = \pi r^2\ 입니다')

# 캡션

st.caption('짧은 설명')

# 이모티콘 출력

"썬글라스 -> :sunglasses:"
"붐업 -> :+1:"
"웃음 -> :smile:"
```

# 2. data 출력 방식
- df: 매직커맨드 
- st.write(): 텍스트 뿐만 아니라 객체도 출력 가능 
- st.dataframe(df): 데이터프레임 출력
- st.table(df): 데이터프레임 출력 (가로로 넓게 보이고, True/False 값이 문자 그대로 보임)
- st.json(): JSON 형식으로 출력(웹에서 많이 활용하는 딕셔너리와 유사한 형태의 데이터 포맷)
- st.metric(): 특정 데이터의 상태 변화를 보기 쉽게 출력 

```python
%%writefile my_web/02_data.py
import streamlit as st
import pandas as pd

st.title('📌Data 출력 방식')
""
""

# df생성

df = pd.DataFrame({
    'col1':[1,2,3],
    'col2':['a','b','c'],
    'col3':[True,False,True]
})

df
""

st.write(df)

st.table(df)

st.json([
    {'name':'희윤', 'age':26, 'gender':'female'},
    {'name':'재찬', 'age':26, 'gender':'male'}
])

st.metric(label='온도', value='10°c', delta='1.2') # alt + 0176
st.metric(label='삼성전자', value='61,000', delta="+1,000")
st.metric(label='삼성전자', value='61,000', delta='-1,200')
st.metric(label='현재 사용자', value='25,000')
```

# 3. 미디어 입출력 

- `with`예약어:
    - 파이썬 리소스를 안전하게 관리하기 위한 예약어 (자원을 자동으로 close해주는 예약어)
    - 파일, DB와 처럼 열고 -> 쓰고  -> 반드시 닫아야하는 대상에 쓰임 
    - 위와 같은 대상을 close해주지 않으면 다른 프로세스와 충돌이 되거나 데이터가 꼬이거나 일부만 저장될 위험이 있음 
        - 데이터가 버퍼(메모리)에 남아 프로그램에 적용되지 않을 수도 있음 
        - ex) 파일을 열어서 새로운 내용을 작성해도 메모리에만 남아있고 실제로 파일에는 작성되지 않을 수도 있음 
    - 따라서 close를 통해 실제로 파일에 작성되도록 해줘야 함 
    - with를 통해 매 번 안전하게 close를 해주어 os에서 해당 자원을 해제시켜주도록 함

- `open()` 함수:
    - 파일을 읽거나 쓰거나 수정하기 위해 file 객체를 생성해주는 함수 
    - 주요 매개변수:
        - r: 읽기
        - w: 쓰기
        - a: 이어쓰기 
        - b: 바이너리 모드 

# 5. 다양한 위젯 

- button
    - st.button("제출", key="button3", on_click=my_button, args=(1, 2, 3, 4)): \
    on_click=my_button으로 설정된 함수는 Streamlit 내부의 별도 이벤트 처리 루프에서 호출 됨\
    이 루프는 전체 스크립트 실행 전 또는 중간에 실행될 수 있으며, 출력 위치는 스크립트 흐름과 무관하게 상단부터 채워지게 됨 \
    그래서 결과는 버튼 아래가 아닌 앱의 상단 쪽(스크립트 흐름상 앞부분)에 뜨는 것처럼 보임 
- checkbox
- radio: 라디오 선택 박스(두 개중 하나 선택)
- selectbox: 드롭다운식 선택 박스
- multiselect: 드롭다운식 선택 박스(여러개 값 중복 선택 가능)
- color_picker: 컬러 선택
- select_slider: 여러 개 중 한 가지 선택 슬라이드
- slider: 최대 최소값 사이의 한 개 숫자 선택 
- number_input: 사용자가 입력한 숫자 받기 
- text_input: 사용자가 입력한 텍스트 받기
- date_input: 사용자가 입력한 날짜 받기
- time_input: 사용자가 입력한 시간 받기

# 4. 코드 출력 방식
- st.code(): 코드 출력 (실행 결과 출력 x)
- st.echo(): 코드 출력 + 실행결과 동시에 출력 
- streamlit에서의 with 예약어:
    - 자원 관리 목적으로 쓰이지 않고 레이아웃 관리를 위해 쓰임
    - UI 컴포넌트가 출력될 컨테이너나 영역을 지정하는 컨텍스트 역할 
    - 특정 구간 안에 컴포넌트들이 렌더링되도록 함
    - 즉, 컴포넌트를 묶어주는 개념 

```python
%%writefile my_web/04_code.py
import streamlit as st


st.header('아래는 `.code()` 함수에 대한 예시', divider='yellow')

my_code = """\
st.write("1. `.code()` 예제1")
    total = 0
    for i in range(5):
        total += 1
    print(total)\

"""

my_code2 = """\
    def my_functions():
        print('hellow')\
"""

st.code(my_code)
st.code(my_code2)


# 예제2

st.write('1. `.code()` 예제2')
def get_user_name():
    return 'John'
def get_punctuation():
    return '!!!'

st.code(get_user_name())
st.code(get_punctuation())

st.header('아래는 `.echo()` 함수에 대한 예시', divider='yellow')

st.write('1. 아래 부분의 함수 정의부는 화면에 출력되지 않음')

st.write('1. `.code()` 예제2')
def get_user_name():
    return 'John'
def get_punctuation():
    return '!!!'

user_name = get_user_name()
punctuation = get_punctuation()
greeting = 'Hi, there!'

st.write(user_name, punctuation)

st.write("--DONE!--")
""
""

# 예제2
st.write("2. 아래 코드는 함수의 정의부까지 모두 화면에 출력됨")
with st.echo():
    # 이 코드블럭 내의 모든 코드는 화면에 출력되고 실행됨
    user_name = get_user_name()
    punctuation = get_punctuation()
    greeting = 'Hi, there!'

    st.write(user_name, punctuation)
    
    
```

```python
%%writefile my_web/05_widget.py
import streamlit as st
import pandas as pd
import datetime
import time

st.subheader('버튼', divider='yellow')
st.button(label='클릭', key='btn1')
st.button(label='클릭', key='btn2')

def my_btn(*args):
    total = sum(args)
    st.write(f"합계: {total}")
    
# 버튼 동작 방법1: on_click 매개변수 사용
st.button('제출1', key='btn3', on_click=my_btn, args=(1,2,3,4))

# 버튼 동작 방법2: 조건문 사용
if st.button('제출2', key='btn4'):
    my_btn(1,2,3,4)
    
    
st.subheader('체크박스', divider='yellow')

st.checkbox('위 내용에 동의합니다')

check_box = st.checkbox('위 내용에 동의합니다2')

st.write(check_box) # 반환값: bool 체크시 True

st.subheader('라디오 박스', divider='yellow')

st.radio(label='다음 중 한 가지를 선택하시오 1', options=['통장에 매일 5만원 입금', '지금 바로 2억원 입금'], key='radio')

select_result = st.radio(label='다음 중 한 가지를 선택하시오 2', options=['통장에 매일 5만원 입금', '지금 바로 2억원 입금'], key='radio2')

st.write(select_result)




# ------------------------------------------------

st.subheader('선택박스', divider='yellow')

st.selectbox(label='다음 중 한 가지를 선택하시오 1', options=['통장에 매일 5만원 입금', '지금 바로 2억원 입금'], key='radio3')
select = st.selectbox(label='다음 중 한 가지를 선택하시오 1', options=['통장에 매일 5만원 입금', '지금 바로 2억원 입금'], key='radio4')


st.write(select)


# ------------------------------------------------

st.subheader('다중선택박스', divider='yellow')

st.multiselect(label='오늘 야식은?', options=['치킨', '피자', '햄버거'], key='multi1')
mul_select = st.multiselect(label='오늘 야식은?', options=['치킨', '피자', '햄버거'], key='multi2')


st.write(mul_select)

# -----------------------------

st.subheader('컬러선택', divider='yellow')

st.color_picker('색상을 선택하세요 1', key='color_pic1')
c = st.color_picker('색상을 선택하세요 2', key='color_pic2')
st.write(f'색상 결과 : {c}')




""
#----------------------------------

st.subheader('슬라이더', divider='yellow')

slider = st.slider('추천 점수', min_value=0, max_value=10, value=5, step=2)
st.write(f'slideer 결과: {slider}')
st.write(type(slider))



#----------------------------------

st.subheader('숫자 입력', divider='yellow')

num_input = st.number_input('점수', min_value=0, max_value=100, value=10, step=10)
st.write(num_input)

#----------------------------------

st.subheader('각종 데이터 입력', divider='yellow')

text = st.text_input('이름:')
date = st.date_input('생일:', max_value=datetime.datetime.now())
time_input = st.time_input('태어난 시간:', step=datetime.timedelta(hours=1))

st.write(f'이름:{text}, 생일:{date}, 시간:{time_input}')



#----------------------------------
import time 

st.subheader('진행 상태 바', divider='yellow')

st.progress(30, text='30%')

progress_bar = st.progress(0)

if st.button('진행', key='btn5'):
    for i in range(0,101,25):
        progress_bar.progress(i, text=f'{i}%')
        time.sleep(1)
        
        
        
# ----------------------------------------------------
st.subheader('spinner', divider='yellow')

if st.button('진행', key='btn6'):
    with st.spinner('실행중...'):
        time.sleep(3)
        st.success('종료', icon='✅')
        
        
        
#----------------------------------------------------------
st.subheader('상태메세지', divider='yellow')
st.success('성공!', icon='✅')
st.error("오류 메세지:", icon='😒')
st.warning('경고!', icon='⚠️')
st.info('정보: ', icon='📋')
```

```python
%%writefile my_web/basic_practice.py
import streamlit as st
import pandas as pd
import time

st.title('나만의 작고 소중한 대시보드')
st.header('실습 예제 작성중!')
st.write('이 대시보드는 streamlit을 활용하여 만들어졌습니다.')
st.markdown('이곳에서는 다양한 *Streamlit* 함수를 활용할 수 있습니다. [여기](https://docs.streamlit.io/)를 클릭하세여 더 알아보세요.')

"---"

data_df = pd.DataFrame({
    "상품": ["A", "B", "C", "D"],
    "판매량": [100, 150, 200, 50],
    "작년대비 증감비": ["+10%", "+5%", "+7%", "-15%"]
})

st.table(data_df)
max_row = data_df.loc[data_df['판매량'].idxmax()]
max_sales = max_row['판매량']
max_delta = max_row['작년대비 증감비']

st.metric(label='최대판매량', value=max_sales, delta=max_delta)

select = st.selectbox(label='어떤 상품을 보시겠습니가?', options=data_df['상품'], key='product')

selected_data = int(data_df.loc[data_df['상품'] == select, '판매량'].iloc[0])
st.write(f'{select}의 판매량은 {selected_data}입니다.')

st.markdown("---")
st.latex(r"l = 2\pi r")

"---"
st.header('냥이!')
st.image("../../data/hamster.jpg", caption='냥이 이미지', width=200)

"---"
code_basic = """\
    print('Hello, Streamlit!')\
"""

st.code(code_basic)
st.warning('이 텍스트는 경고용 메시지입니다')

"---"
if st.button(label='성공 버튼', key='success_btn'):
    st.success('성공!')
    
"---"
name_input = st.text_input('이름을 입력하세요:')
st.write(f'안녕하세요, {name_input}님!')

"---"
st.write('색상을 선택하세요')
color_basic = st.color_picker('색상을 선택하세요', key='color_pic_basic')
st.write(f'선택한 색상은 {color_basic}입니다.')

"---"
if st.button('진행버튼', key='prog_btn'):
    progress_bar = st.progress(0)
    for i in range(0,101,10):
        progress_bar.progress(i)
        time.sleep(1)
        

```
